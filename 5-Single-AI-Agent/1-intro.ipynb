## AI Agent 🤖

An **AI Agent** is an intelligent system that:

- Receives a **high-level goal** from the user.
- **Plans and decides** what actions to take.
- **Executes** actions using external tools, APIs, or knowledge sources.
- Maintains **context** and reasons over multiple steps.
- **Adapts** to new information.
- Optimizes actions to achieve the intended outcome.

### 🧠 Simple Formula

**AI Agent = LLM + Tools + Context + Reasoning**

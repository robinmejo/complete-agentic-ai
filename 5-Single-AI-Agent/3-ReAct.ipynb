# ReAct (Reasoning + Acting)

## What is ReAct?

**ReAct** stands for **Reasoning + Acting**.

It is a design pattern used for building **AI Agents** where the LLM combines:

- 🧠 **Reasoning (Thought)** — deciding what needs to be done
- 🔧 **Acting (Action)** — using tools to perform an action
- 👀 **Observation** — looking at the result of the action

Instead of generating an answer in one step, the agent can work through a **multi-step process**.

---

# Agent & Agent Executor

## Agent

An **Agent** is the decision-making part of an AI system.

It uses the **LLM** to decide:

- 🧠 What needs to be done
- 🔧 Which tool should be used
- 📝 What input should be given to the tool
- ✅ When the task is complete

The Agent decides **what action to take next**, but it does not necessarily execute the tool itself.

---

## Agent Executor

The **AgentExecutor** is responsible for **running and orchestrating the entire agent loop**.

It connects the Agent with the available Tools and repeatedly executes the Agent's decisions.

### AgentExecutor Loop

```text
User Input
    ↓
Agent
    ↓
Next Action 🔧
    ↓
AgentExecutor
    ↓
Execute Tool
    ↓
Tool Observation 👀
    ↓
Add Observation to History
    ↓
Agent
    ↓
Next Action
    ↓
...
    ↓
Final Answer ✅

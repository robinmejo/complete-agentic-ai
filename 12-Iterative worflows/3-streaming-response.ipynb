# Streaming Response in LangGraph

## What is Streaming?

Normally, an LLM gives the complete response at once.

### Normal response

```text
User: Explain LangGraph

          ↓

LLM thinks

          ↓

"LangGraph is a framework for building..."

Streaming response

With streaming, the response is shown piece by piece as the LLM generates it.

User: Explain LangGraph

          ↓

"LangGraph"

"LangGraph is"

"LangGraph is a"

"LangGraph is a framework"

...

| Stream Mode | What it provides |
|-------------|------------------|
| `"updates"` | State updates from each node |
| `"values"` | Full state after each step |
| `"messages"` | LLM messages/tokens |
| `"custom"` | Custom data emitted from nodes |
| `"debug"` | Detailed execution/debug information |

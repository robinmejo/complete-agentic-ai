# LLM — Common Applications

An **LLM (Large Language Model)** takes a **prompt** as input and generates a **response** based on the context and instructions provided.

### Common Applications of LLMs

#### 1. Translation

Converts text from one language to another.

**Example:**

* Input: `Hello, how are you?`
* Output: `Hola, ¿cómo estás?`

LLMs can translate between many different languages while considering the context of the sentence.

---

#### 2. Summarization

Converts a long piece of text into a shorter version containing the important information.

**Example:**

```text
Long article → LLM → Short summary
```

Useful for:

* Summarizing documents
* News articles
* Meeting notes
* Research papers

---

#### 3. Chat

LLMs can interact with users using natural language.

**Example:**

```text
User → Prompt → LLM → Response
```

This is the foundation of:

* Chatbots
* AI assistants
* Customer support systems
* Coding assistants

---

#### 4. NER — Named Entity Recognition

Identifies important entities in text and classifies them into categories.

Common entity types include:

* **PERSON**
* **LOCATION**
* **ORGANIZATION**
* **DATE**
* **PRODUCT**

**Example:**

```text
"Robin works at Infosys in Pune."

PERSON       → Robin
ORGANIZATION → Infosys
LOCATION     → Pune
```

---

### Key Idea

```text
                 ┌─────────┐
Prompt ────────► │   LLM   │ ────────► Response
                 └─────────┘
                      │
        ┌─────────────┼─────────────┐
        ↓             ↓             ↓
   Translation   Summarization    Chat
                      │
                      ↓
                     NER
```

**Remember:** An LLM is a general-purpose language model that can perform many language-related tasks depending on the prompt and instructions.

## Why RAG Was Introduced

**RAG (Retrieval-Augmented Generation)** was introduced to overcome some limitations of LLMs:

* **Knowledge limitation** → LLMs may not know private or recent information.
* **Hallucination** → LLMs can generate incorrect information.
* **Outdated knowledge** → LLMs may not have the latest information.
* **Private data** → LLMs need access to company-specific documents and data.

### RAG Approach

**User Query → Retrieve Relevant Information → LLM + Context → Response**

> **RAG connects an LLM with external knowledge to generate more accurate, relevant, and up-to-date responses.**

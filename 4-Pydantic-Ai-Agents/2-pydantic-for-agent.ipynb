```python
from pydantic import BaseModel,EmailStr,AnyUrl, Field
from typing import Dict, List,Optional,Annotated

class PatientChildClassData(BaseModel):
    patient_id: str
    ''' What does Annotated do?
    It lets you attach extra information/metadata to a type. 
    So think like Annotated[type, extra information]'''
    # name: str=Field(max_length=50, description="The name of the patient") # Field is a Pydantic class that allows you to add metadata to your model fields.
    name: Annotated[
        str,
        Field(max_length=50, description="The name of the patient")
    ]
    age: int= Field(gt=0,lt=120, description="The age of the patient") # gt and lt are Pydantic validators that check if the value is greater than or less than a certain value.
    gender: str
    diagnosis: str
    email:Optional[EmailStr] = None # EmailStr is a Pydantic type that validates email addresses.
    weight: Annotated[float, Field(gt=0, strict=True, description="The weight of the patient in kg")]
    allergies: Optional[List[str]] = Field(default=None, description="List of allergies",max_items=10)
    ''' List is a type hint in Python that means a collection of multiple values.
    This means:
    names should be a list of strings.'''
    contact_info: Dict[str, str]
    ''' List and Dict from typing are the older style of type hinting:
    🧠 Think:
    Old style: List[str]
    Modern style: list[str] ✅ after python 3.9 '''
    linkedin_profile: Optional[AnyUrl] = None # AnyUrl is a Pydantic type that validates URLs.
    

def add_patietnt_data(patient: PatientChildClassData):
    return f"Patient {patient.name} with ID {patient.patient_id} has been added to the system."

patient_data={
    "patient_id": "P12345",
    "name": "John Doe",
    "age": 30,
    "gender": "Male",
    "diagnosis": "Flu",
    "allergies": ["Peanuts", "Shellfish"],
    "contact_info": {"phone": "123-456-7890", "email": "john.doe@example.com"},
    "weight": 70.5
}

patient = PatientChildClassData(**patient_data) # is used to convert the dictionary into a Pydantic object.

# The ** unpacks the dictionary

added_patient_data = add_patietnt_data(patient)
added_patient_data


```

# Pydantic

- **Pydantic** is a Python library used for **data validation and parsing**.
- It uses **Python type hints** to define the expected data structure.
- Mainly used in **APIs, JSON data, and FastAPI**.

## Basic Example

```python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

user = User(name="Robin", age=25)

print(user)

### Key Points

- `BaseModel` → creates a Pydantic model
- **Type hints** → define expected data types , example (: str)
- **Validation** → automatically validates input
- `model_dump()` → converts model to a dictionary

**Pydantic = Type Hints + Validation + Parsing**

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

user = User(name="Robin", age=25)

print(user.name+" is " + str(user.age) + " years old.")

print(user)
user.model_dump()

from langchain import PromptTemplate
````

# 🤖 Multi-Agent Research Assistant

A **Multi-Agent Research Assistant** is an autonomous AI system where multiple specialized agents collaborate to create a research report.

## Agents

- 🔎 **Search Agent** → Finds relevant and recent sources.
- 📖 **Reader Agent** → Reads sources and extracts useful information.
- ✍️ **Writer Agent** → Creates a structured research report.
- 🧐 **Critic Agent** → Reviews, scores, and provides feedback.

## Workflow

**User Question → Search → Read → Write → Critic → Final Report**

## Technologies

- **LangChain** → Builds and orchestrates the agents.
- **LCEL** → Connects components into a pipeline.
- **Shared Memory** → Allows agents to share information and work together.

## 🧠 Remember

**Search → Read → Write → Critic**

> Multiple specialized agents + shared memory + orchestration = **Multi-Agent System**

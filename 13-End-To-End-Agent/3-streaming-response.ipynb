### Normal response

```text
User: Explain LangGraph

          ↓

LLM thinks

          ↓

Streaming response

With streaming, the response is shown piece by piece as the LLM generates it.

          ↓

"LangGraph"

"LangGraph is"

"LangGraph is a"

"LangGraph is a framework"

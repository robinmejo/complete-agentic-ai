# LangChain Overview

## What is LangChain?

- **Open-source framework** for building LLM-powered applications.
- Provides reusable components to connect **LLMs, prompts, data, tools, and workflows**.
- Supports multiple LLM providers through a common interface.

---

## Core Components

| Component | Technical Purpose |
|---|---|
| **Models** | Interface with LLMs such as OpenAI, Anthropic, Gemini, etc. |
| **Prompts** | `PromptTemplate`, `ChatPromptTemplate` for structured prompts |
| **Output Parsers** | Convert LLM output into structured formats |
| **Retrievers** | Retrieve relevant documents from a data source/vector store |
| **Chains / Runnables** | Connect multiple processing steps |
| **Tools** | Functions/APIs that an LLM can invoke |
| **Agents** | LLM decides which tools/actions to execute |

---

## Common Applications

- **Chatbots**
- **RAG (Retrieval-Augmented Generation)**
- **Document Q&A**
- **Text Summarization**
- **Multistep LLM Workflows**
- **Tool-using Agents**

---

## RAG Flow

`Documents → Embeddings → Vector Store → Retriever → LLM → Response`

---

## Agent Flow

`User Input → Agent → Tool Selection → Tool Execution → LLM → Response`

---

## Key Concepts

- **LLM** → Generates text/reasoning
- **Prompt** → Instructions given to the LLM
- **Retriever** → Fetches relevant information
- **Chain** → Fixed sequence of operations
- **Agent** → Dynamically decides what action/tool to use
- **Tool** → External function/API available to the agent

> **LangChain = Components + Runnables + Tools + Agents → LLM Applications**

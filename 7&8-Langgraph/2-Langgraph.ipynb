# Fault Tolerance

**Fault Tolerance** is the ability of a system to continue operating correctly even when one or more components fail.

### Key Techniques

- **Retry** → Re-execute a failed operation.
- **Exponential Backoff** → Increase the delay between retries to avoid overloading the service.
- **Timeout** → Stop waiting after a defined time limit.
- **Fallback** → Use an alternative service/component when the primary fails.
- **Circuit Breaker** → Temporarily stop requests to a failing service.
  - `CLOSED` → Normal operation
  - `OPEN` → Block requests
  - `HALF-OPEN` → Test recovery
- **Redundancy** → Maintain multiple instances of a component.
- **Replication** → Maintain multiple copies of data/services.
- **Failover** → Automatically switch from a failed component to a backup.
- **Graceful Degradation** → Continue with reduced functionality instead of complete failure.
- **Health Checks** → Continuously verify service/component health.
- **Checkpointing** → Save intermediate state so processing can resume after failure.

### AI / Agentic AI

For LLM/agent systems, fault tolerance can be implemented using:

`Retry + Timeout + Fallback + Circuit Breaker + Checkpointing + Error Handling`

**Goal:** Avoid a single component failure causing the entire workflow to fail.

> **Interview Definition:** Fault tolerance is the ability of a system to continue functioning despite failures in its components or dependencies.

# LangGraph Overview

## What is LangGraph?

**LangGraph** is an orchestration framework for building **stateful, multi-step, and event-driven workflows** with LLMs.

Think of LangGraph as a **flowchart engine for LLM applications**:

- **Nodes** → Steps/tasks in the workflow
- **Edges** → Connections between steps
- **State** → Shared data maintained throughout the workflow
- **Conditional Edges** → Dynamic routing/branching

### Key Features

- State management
- Conditional branching
- Loops
- Human-in-the-loop
- Pause & resume
- Fault recovery
- Single-agent workflows
- Multi-agent coordination
- Event-driven/asynchronous execution

## LangChain vs LangGraph

**LangChain** → Best for **simple, mostly linear workflows**

Examples:
- Prompt chains
- Summarization
- Basic RAG

**LangGraph** → Best for **complex, non-linear workflows**

Use when you need:

- Conditional paths
- Loops
- Human-in-the-loop
- Multi-agent coordination
- Asynchronous/event-driven execution

> **Easy way to remember:**  
> **LangChain = chain of steps**  
> **LangGraph = graph of steps with state, branching, and loops**

# LangChain + LangGraph

## Should We Still Use LangChain?

**Yes.** LangGraph is built on top of **LangChain** and does **not replace it**.

### LangChain Provides Building Blocks

Common LangChain components:

- **ChatOpenAI / LLMs** → Interact with language models
- **PromptTemplate** → Create structured prompts
- **Retrievers** → Retrieve relevant information
- **Document Loaders** → Load documents/data
- **Tools** → Enable agents to perform actions

### Key Difference

**LangChain** → Provides the **building blocks**

**LangGraph** → Handles the **workflow orchestration**

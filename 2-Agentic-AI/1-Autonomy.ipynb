## 🤖 Autonomy in Agentic AI

In **Agentic AI**, **autonomy** means the AI can **independently decide what actions to take to achieve a goal**, without needing the user to tell it every step.

### 💡 Simple Example

**🧠 Normal LLM:**

> 👤 User: "Find the best flight."
>
> 🤖 AI: Gives an answer.

**🤖 Agentic AI with Autonomy:**

> 👤 User: "Plan my trip to Delhi."
>
> The AI can independently decide to:
>
> 1. ✈️ **Search flights**
> 2. 💰 **Compare prices**
> 3. 🏨 **Search hotels**
> 4. 🗓️ **Create an itinerary**
> 5. ✅ **Ask for user approval before booking**

### 🎯 Key Idea

**Autonomy = Ability of an AI agent to independently decide and act toward a goal.**

**Goal → 🧠 Plan → 🤔 Decide → ⚙️ Act → 👀 Observe → 🔄 Repeat**

### 4. Autonomy can be controlled

**a. Permission Scope** – Limit what tools or actions the agent can perform independently.  
> 💡 Example: The agent can screen candidates, but needs approval before rejecting anyone.

**b. Human-in-the-Loop (HITL)** – Insert checkpoints where human approval is required before continuing.  
> 💡 Example: The agent asks, **"Can I post this JD?"** before publishing it.

**c. Override Controls** – Allow users to stop, pause, or change the agent's behaviour at any time.  
> 💡 Example: Update the screening command to halt resume processing.

**d. Guardrails / Policies** – Define hard rules or ethical boundaries the agent must follow.  
> 💡 Example: **Never schedule interviews on weekends.**

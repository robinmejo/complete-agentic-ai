## 1. 🧠 Brain

The **Brain** is the intelligence/reasoning component of an AI agent. It is usually an **LLM (Large Language Model)**.

Technically, the Brain is responsible for:
- Understanding the user's goal
- Reasoning about the task
- Deciding what action to take
- Selecting the appropriate tool
- Interpreting tool results
- Generating the final response

Common LLMs:
- OpenAI GPT
- Anthropic Claude
- Google Gemini
- Meta Llama

In **LangChain**, the Brain can be represented using a **Chat Model** such as `ChatOpenAI` or `ChatGroq`.

**Example:**

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o")

## 2. 🎯 Orchestrator

The **Orchestrator** manages the agent's workflow and decides what should happen next.

Technically, it can be implemented using:
- **LangGraph** → Graph-based workflow orchestration
- **State** → Stores the current workflow information
- **Nodes** → Individual steps/agents
- **Edges** → Define how execution moves between nodes

**Example:**

User Request → LLM → Tool → Result → LLM → Final Answer

**Key idea:**  
The Orchestrator controls **which step runs next and when the task is complete**.

## 3. 🛠️ Tools

**Tools** give an AI agent the ability to interact with external systems.

Technically, tools can be:
- Python functions
- REST APIs
- Database queries
- Web search
- File/PDF processing
- External services

In **LangChain**, tools can be created using `@tool`.

**Example:**

```python
from langchain.tools import tool

@tool
def add(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b


## 4. 🧠 Memory

**Memory** allows an agent to remember and retrieve information.

Technically, memory can be implemented using:
- **State** → Current conversation/task information
- **Checkpoints** → Save agent state during execution
- **Vector Database** → Store and retrieve long-term information
- **Embeddings** → Convert text into numerical representations for semantic search

**Examples:**
- Short-term → Conversation history
- Long-term → User preferences stored in a vector database

In **LangGraph**, checkpoints can be used to persist agent state.

**Key idea:**  
Memory allows the agent to **remember information and use it in future steps**.

## 5. 👨‍💼 Supervisor

A **Supervisor** manages and coordinates multiple specialized agents.

Technically, the Supervisor is often an **LLM-based router/controller** that decides which agent should handle the next task.

**Example:**

```text
                 Supervisor
                     ↓
          ┌──────────┼──────────┐
          ↓          ↓          ↓
     Research      Travel     Finance
       Agent        Agent       Agent

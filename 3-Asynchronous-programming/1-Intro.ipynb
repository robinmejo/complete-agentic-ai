## ⚡ Asynchronous Programming in Python

**Asynchronous programming** allows a program to handle multiple tasks **concurrently without blocking execution**.

It is mainly useful for **I/O-bound tasks**, where the program spends time waiting for external operations such as:

- 🌐 Network/API requests
- 📁 File I/O
- 🗄️ Database queries
- 🔌 External services

### 🔑 Key Concepts

- `async` → Defines an asynchronous function (coroutine)
- `await` → Waits for an async operation without blocking other tasks
- `asyncio` → Python's built-in framework for asynchronous programming
- `asyncio.gather()` → Runs multiple async tasks concurrently

### Example

```python
import asyncio

async def task(name):
    print(f"Starting {name}")
    await asyncio.sleep(2)
    print(f"Finished {name}")

async def main():
    await asyncio.gather(
        task("Task 1"),
        task("Task 2")
    )

await main()

# 🔄 Subroutine vs Coroutine

## 📌 Subroutine
A **subroutine** executes from start to finish and then returns control to the caller.

**Flow:**  
▶️ Call → ⚙️ Execute → ↩️ Return

### 🐍 Python
```python
def greet():
    print("Hello")
    print("Goodbye")

greet()


## ⏸️ Coroutine
A **coroutine** coroutine can pause its execution and later resume from the same point.

**Flow:**  
▶️ Start → ⏸️ Suspend → 🔄 Resume → ✅ Finish

### 🐍 Python
```python
import asyncio

async def task():
    print("Started")
    await asyncio.sleep(1)  # Pause
    print("Resumed")

await task()

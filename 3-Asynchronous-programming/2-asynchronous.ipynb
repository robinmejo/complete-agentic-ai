Synchronous programming

````python
import time 

def fetch_weather_data():
    print("Fetching weather data...")
    time.sleep(2)  # Simulate delay
    print("Weather data fetched successfully.")

def fetch_news_data():
    print("Processing news data...")
    time.sleep(3)  # Simulate delay
    print("News data processed successfully.")
````

````python
def main():
    print("Inside main function")

start_time = time.time()

fetch_weather_data()
fetch_news_data()

end_time = time.time()
print(f"Total time taken: {end_time - start_time} seconds")

main()
````

Asynchronous programming

````python
import asyncio

import time 

async def fetch_weather_data():
    print("Fetching weather data...")
    await asyncio.sleep(2)  # Simulate delay
    print("Weather data fetched successfully.")

async def fetch_news_data():
    print("Processing news data...")
    await asyncio.sleep(3)  # Simulate delay
    print("News data processed successfully.")

async def main():
    print("Inside main function")

start_time = time.time()

await asyncio.gather(fetch_weather_data(), fetch_news_data())

end_time = time.time()
print(f"Total time taken: {end_time - start_time} seconds")

await main()
````

⭐ Key points

| Concept            | Meaning                                                         |
| ------------------ | --------------------------------------------------------------- |
| `async def`        | Defines an asynchronous/coroutine function                      |
| `await`            | Waits for an async operation without blocking other async tasks |
| `asyncio.sleep()`  | Simulates an asynchronous delay                                 |
| `asyncio.gather()` | Runs multiple coroutines concurrently and waits for all         |
| `asyncio.run()`    | Starts the asyncio event loop and runs the main coroutine       |
| `time.time()`      | Used to measure execution time                                  |
| Event Loop         | Manages and schedules asynchronous tasks                        |


## 🧠 Easy way to remember

### Concurrency

**Concurrency = dealing with multiple tasks at the same time by switching between them.**

```text
Task A: ███ waiting ███
Task B:    ███ waiting ███
Task A:       ███
Task B:          ███
```
### Parallelism

**Parallelism = multiple tasks actually execute at the same time.**

```text
CPU Core 1 → Task A █████████
CPU Core 2 → Task B █████████
CPU Core 3 → Task C █████████
CPU Core 4 → Task D █████████
